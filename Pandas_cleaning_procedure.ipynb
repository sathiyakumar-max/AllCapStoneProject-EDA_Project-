```python
import pandas as pd
import matplotlib.pyplot as plt # used for visual ananlysis
import seaborn as sns  # used for visual ananlysis
my_csv = pd.read_csv("D:\\Data Analyst Course\\Train.csv",sep = ',') #data loading completed
```

```python
plt.hist(df['Item_MRP'],bins=10 ,color='red') # histogram single variable analysis used rarely
plt.xlabel('Item_MRP')
#plt.ylabel('Item_MRP')
plt.title('Analysis of Item_MRP')
plt.show()
```

```python
sns.boxplot(x=df['Item_Outlet_Sales']) #boxplot
plt.show()
```

```python
sns.countplot(x=df['Item_Outlet_Sales'],data=df) # countplot
plt.xticks(rotation =90)
plt.show()
```

```python
df['Item_MRP'].value_counts().plot.pie(autopct='%2.5f%%')
plt.show()
```

```python
sns.regplot(x=df['Outlet_Establishment_Year'],y=df['Item_Outlet_Sales'],scatter = True,line_kws={"color":"red"}) #Bi variant Analysis -scatter plot 
#for continous +continuous data
plt.title("Outlet_Establishment_Year vs Item_Outlet_Sales")
plt.xlabel("Outlet_Establishment_Year")
plt.ylabel("Item_Outlet_Sales")
plt.show()
```

```python
sns.boxplot(x=df['Item_Type'],y=df['Item_Outlet_Sales']) #boxplot for bivarient Analysis Catagerized data + continuous data
plt.xticks(rotation = 90)
plt.show()
```

```python
sns.violinplot(x=df['Item_Type'],y=df['Item_Outlet_Sales']) #violinplot same as boxplot display chages
plt.xticks(rotation = 90)
plt.show()
```

```python
#categorical data + categorical data
sns.countplot(x='Item_Type',hue ='Outlet_Size',data=df) 
plt.title('Item_Type vs Outlet_Size')
plt.xlabel('Item_Type')
plt.ylabel('Outlet_Size')
plt.xticks(rotation = 90)
plt.show()
```

```python
# bar chart can also be brought using crosstab in Pandas
ct=pd.crosstab(df['Item_Type'],df['Outlet_Size'])
ct.plot(kind='bar',stacked=True,figsize=(10,5))
plt.title('Item_Type vs Outlet_Size')
plt.xlabel('Item_Type')
plt.ylabel('Outlet_Size')
plt.xticks(rotation = 90)
plt.show()
```

```python
# Multivarient analysis - more than 2 variables
# Multivarient mixed up with both catagorical and continuous then we need to go with pairplot
sns.pairplot(df[['Item_Weight','Item_Visibility','Item_MRP','Item_Outlet_Sales']]) #pairplot is for numeric data
plt.show()
```

```python
# Multivarient mixed up with both catagorical and continuous then we need to go with heatmap
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()
```

```python
# Multivarient with 2 categorial + 1 continuous - box plot
sns.boxplot(x='Item_Type',y='Item_Outlet_Sales',hue ='Outlet_Size',data=df)
plt.xticks(rotation =90)
plt.show()
```

```python
# Multivarient with 2 categorial + 1 continuous - barplot can also used
sns.barplot(x='Item_Type',y='Item_Outlet_Sales',hue ='Outlet_Size',data=df)
plt.xticks(rotation =90)
plt.show()
```

```python
# Multivarient with 2 continuous + 1 categorial - scatterplot can also used
sns.scatterplot(x='Item_MRP',y='Item_Outlet_Sales',hue ='Outlet_Size',data=df)
plt.xticks(rotation =90)
plt.show()
```
